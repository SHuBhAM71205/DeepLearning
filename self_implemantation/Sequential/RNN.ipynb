```python
class Unit:
    def __init__(self,input,state):
        self.input=input
        self.state=state
        
    
    def get_output(self,prev_state):
        self.output
        
        
        return self.output
    
    def get_state(self):
        return self.state
    
class RNN:
    def __init__(self,input_shape,output_shape,state_shape):
        self.input_shape=input_shape
        self.output_shape=output_shape
        self.state_shape=state_shape
        
        self.__init_weight(input_shape,output_shape,state_shape)
        self.__init_bias(input_shape,output_shape,state_shape)
        
        self.state = np.zeros(state_shape)
        
        self.unit_list:List[RNN]=[]
    
    def __init_weight(self,input_shape,output_shape,state_shape):
        ## initialize with zavier which is underroot of the
        self.W=np.random.randn(state_shape, state_shape) * np.sqrt(1.0 / state_shape)
        
        self.U=np.random.randn(input_shape,state_shape) * np.sqrt(1.0 / input_shape)
        
        self.V=np.random.randn(state_shape,output_shape) * np.sqrt(1.0 / state_shape)
        
    def __init_bias(self,input_shape,output_shape,state_shape):
        ## initialize bias witht the zeros
        self.b_in=np.zeros((1, input_shape))
        
        self.b_state= np.zeros((1, state_shape))
        
        self.b_out =  np.zeros((1, output_shape))
        
    def forward(self, x_input):
        
        self.preactivation_state = []
        self.activation_state = [self.state]
        self.preactivation_out = []
        self.activation_out = []

        for x in x_input:
            h_pre = self.W.T @ self.activation_state[-1] + self.U.T @ x + self.b_state.flatten()
            self.preactivation_state.append(h_pre)

            h = self.sigmoid(h_pre)
            self.activation_state.append(h)

            y_pre = self.V.T @ h + self.b_out.flatten()
            self.preactivation_out.append(y_pre)

            y = self.softmax(y_pre[np.newaxis, :])[0]
            self.activation_out.append(y)

        return np.array(self.activation_out)

        
    def backward(self):
        pass
    
    def loss(self):
        loss_t=[]
        
        return loss_t
    
    def reset(self):
        self.unit_list=[]
        
    def sigmoid(self,x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self,x):
        s = self.sigmoid(x)
        return s * (1 - s)

    
    def softmax(self,x):
        exp_x = np.exp(x - np.max(x, axis=1, keepdims=True))
        return exp_x / np.sum(exp_x, axis=1, keepdims=True)
```
